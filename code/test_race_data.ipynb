```python
import pandas as pd
import os

# Load NASCAR race data from the project's data folder
data_path = '../data/external/nascar_race_data_full.csv'

df = pd.read_csv(data_path, encoding='latin-1', low_memory=False)

print(f"Successfully loaded {len(df):,} race records!")
print(f"\nColumns available:")
print(list(df.columns))
```

## Explore the Data Structure

```python
# Preview the data
print("First 10 rows:")
print(df.head(10))

print("\n" + "="*50)
print("\nData types:")
print(df.dtypes)

print("\n" + "="*50)
print("\nBasic statistics:")
print(df.describe())
```

## Filter for Recent Races

```python
# Filter for recent Cup Series races (2023-2024)
# The date column is called 'Date' (not 'race_date') and needs parsing
df['Date'] = pd.to_datetime(df['Date'])
recent_races = df[df['Date'] >= '2023-01-01'].copy()

print(f"Recent races (2023+): {len(recent_races):,} records")

# Show unique tracks
if 'Site' in recent_races.columns:
    print(f"\nTracks in recent data:")
    print(recent_races['Site'].value_counts().head(10))

# Show race winners (if available)
if 'Driver' in recent_races.columns:
    print("\nSample of recent race results:")
    print(recent_races[['Date', 'Driver', 'Pos', 'Site']].head(20))
```

```python
import os

# Create directory if it doesn't exist
os.makedirs('data/raw', exist_ok=True)
os.makedirs('data/external', exist_ok=True)

# Save the full dataset for later use
df.to_csv('data/external/nascar_race_data_full.csv', index=False)
print(f"Full dataset saved to data/external/nascar_race_data_full.csv")

# Save a filtered sample for testing
sample = df.tail(1000)  # Last 1000 records
sample.to_csv('data/raw/test_race_results.csv', index=False)
print(f"Test sample saved to data/raw/test_race_results.csv ({len(sample)} rows)")

print("\n✓ Race data test complete!")
```
